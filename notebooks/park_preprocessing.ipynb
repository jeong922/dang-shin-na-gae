```python
import pandas as pd
from pathlib import Path
```

```python

df = pd.read_csv("../data/raw/서울시 주요 공원현황.csv", encoding="cp949")

print(f"공원 개수 : {len(df)}")

df.head()
```

```python
# 결측치 확인
df.isnull().sum()
```

```python
target_columns = [
    "공원명",
    "개원일",
    "면적",
    "주요시설",
    "주요식물",
    "이용시참고사항",
    "지역",
    "공원주소",
    "관리부서",
    "전화번호",
    "이미지",
    "X좌표(WGS84)",
    "Y좌표(WGS84)"
]

parks = df[target_columns].copy()
```

```python
parks = parks.rename(columns={
    "공원명": "name",
    "개원일": "opened_at",
    "면적": "area",
    "주요시설": "facilities",
    "주요식물": "plants",
    "이용시참고사항": "notes",
    "지역": "district",
    "공원주소": "address",
    "관리부서": "department",
    "전화번호": "phone",
    "이미지": "image",
    "X좌표(WGS84)": "lon",
    "Y좌표(WGS84)": "lat",
})

parks.head()
```

```python
text_columns = [
    "area",
    "facilities",
    "plants",
    "notes"
]

parks[text_columns] = parks[text_columns].fillna("")

parks = parks.dropna(subset=["lat", "lon"]).reset_index(drop=True)
```

```python
parks["area_size"] = (
    parks["area"]
        .astype(str)
        .str.extract(r"([\d.]+)")
        .fillna(0)
        .astype(float)
)

parks[["area", "area_size"]].head()
```

```python
parks["search_text"] = (
    parks["facilities"]
    + " "
    + parks["notes"]
    + " "
    + parks["plants"]
)
```

```python
FORBIDDEN = [
    "애완견 금지",
    "반려견 금지",
    "반려동물 금지",
    "애완동물 금지",
    "출입 금지",
]

DOG_PARK = [
    "반려견 놀이터",
    "반려견놀이터",
    "반려견 운동장",
    "애완견 놀이터",
]

HILLY = [
    "등산로",
    "오르막",
    "계단",
    "임야",
    "자연 산책로",
    "산책 코스",
]
```

```python
parks["is_dog_forbidden"] = parks["search_text"].apply(
    lambda text: any(keyword in text for keyword in FORBIDDEN)
)

parks["has_dog_park"] = parks["search_text"].apply(
    lambda text: any(keyword in text for keyword in DOG_PARK)
)

parks["has_parking"] = parks["search_text"].str.contains(
    "주차장",
    regex=False
)

parks["is_hilly"] = parks["search_text"].apply(
    lambda text: any(keyword in text for keyword in HILLY)
)

parks["has_restroom"] = parks["search_text"].str.contains(
    "화장실",
    regex=False
)

parks["has_store"] = parks["search_text"].str.contains(
    "매점",
    regex=False
)

parks["has_lawn"] = parks["search_text"].str.contains(
    "잔디",
    regex=False
)

parks["has_water"] = parks["search_text"].str.contains(
    "음수대",
    regex=False
)
```

```python
parks[
    [
        "name",
        "area_size",
        "has_parking",
        "has_restroom",
        "has_lawn",
        "has_water",
        "has_dog_park",
        "is_dog_forbidden",
        "is_hilly",
    ]
].head(10)
```

```python
print(f"전체 공원 : {len(parks)}")
print(f"반려동물 출입 금지 : {parks['is_dog_forbidden'].sum()}")
print(f"반려견 놀이터 : {parks['has_dog_park'].sum()}")
print(f"주차장 : {parks['has_parking'].sum()}")
print(f"화장실 : {parks['has_restroom'].sum()}")
print(f"매점 : {parks['has_store'].sum()}")
print(f"잔디 : {parks['has_lawn'].sum()}")
print(f"음수대 : {parks['has_water'].sum()}")
print(f"경사 예상 : {parks['is_hilly'].sum()}")
```
