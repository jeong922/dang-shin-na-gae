```python
import pandas as pd
import requests
import time

parks = pd.read_csv('../data/processed/parks.csv')
parks.head()
```

```python
from dotenv import load_dotenv
import os

load_dotenv("../.env")

API_KEY = os.getenv("GOOGLE_MAPS_API_KEY")

lat = parks.loc[0, "lat"]
lon = parks.loc[0, "lon"]

print(lat, lon)
```

```python
url = (
    "https://maps.googleapis.com/maps/api/elevation/json"
    f"?locations={lat},{lon}"
    f"&key={API_KEY}"
)

response = requests.get(url)

data = response.json()

print(data)
```

```python
elevation = data["results"][0]["elevation"]

print(elevation)
```

```python
parks.loc[0, "elevation"] = elevation

parks.head()
```

```python
import requests

BATCH_SIZE = 50
elevations = []

for start in range(0, len(parks), BATCH_SIZE):

    batch = parks.iloc[start:start + BATCH_SIZE]

    locations = "|".join(
        batch.apply(
            lambda row: f"{row['lat']},{row['lon']}",
            axis=1
        )
    )

    url = (
        "https://maps.googleapis.com/maps/api/elevation/json"
        f"?locations={locations}"
        f"&key={API_KEY}"
    )

    response = requests.get(url, timeout=10)
    response.raise_for_status()

    data = response.json()

    if data["status"] != "OK":
        raise Exception(f"Elevation API Error: {data['status']}")

    elevations.extend(
        result["elevation"]
        for result in data["results"]
    )

    batch_number = start // BATCH_SIZE + 1
    print(
        f"Batch {batch_number} 완료 "
        f"({len(elevations)}/{len(parks)})"
    )
```

```python
parks["elevation"] = elevations
```
